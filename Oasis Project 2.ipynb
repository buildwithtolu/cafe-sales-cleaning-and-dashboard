```python
for col in ['Item', 'Payment Method', 'Location']:
    pj2[col] = pj2[col].fillna('Unknown')
```

For null values in categorical data in this case was filled with a placeholder text "Unknown" specifying that the category is not known

```python
pj2['Total Spent'] = pj2['Total Spent'].fillna(pj2['Quantity'] * pj2['Price Per Unit'])

pj2['Quantity'] = pj2['Quantity'].fillna(pj2['Total Spent'] / pj2['Price Per Unit'])

pj2['Price Per Unit'] = pj2['Price Per Unit'].fillna(pj2['Total Spent'] / pj2['Quantity'])
```

```python
pj2.info()
```

```python
pj2['Price Per Unit'] = pj2['Price Per Unit'].fillna(pj2.groupby('Item')['Price Per Unit'].transform('median'))
pj2['Price Per Unit'] = pj2['Price Per Unit'].fillna(pj2['Price Per Unit'].median())
```

```python
pj2['Quantity'] = pj2['Quantity'].fillna(pj2['Quantity'].median())
```

```python
pj2['Total Spent'] = pj2['Total Spent'].fillna(pj2['Quantity'] * pj2['Price Per Unit'])
```

Quantity, Price Per Unit and Total Spent: All derived from the other two columns i.e (Total = Quantity * Price). Rows where two of three were missing were filled with medians (for price: median per item, then overall; for quantity: overall median) and then total spent was then recalculated.

```python
text_cols = pj2.select_dtypes(include = 'object').columns

for col in text_cols:
    n_raw = pj2[col].nunique()
    cleaned = pj2[col].str.strip().str.lower().nunique()
    print(col, '| raw:', n_raw, '| cleaned:', cleaned)
```

Checked every text column for inconsistent formatting by comparing the number of unique labels before and after `strip` and `lower`

The counts are identical, so no case or spacing inconsistencies exist and no changes were made.

```python
pj2['Quantity'] = pj2['Quantity'].round().astype(int)
```

Quantity is a count, so it was converted from float to integer. `round()` was applied before `astype(int)`to avoid any floating-point noise being truncated to the wrong number

```python
pj2.info()
```

```python
num_cols = ['Quantity', 'Price Per Unit', 'Total Spent']

for col in num_cols:
    q1 = pj2[col].quantile(0.25)
    q3 = pj2[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    flagged = pj2[(pj2[col] < lower) | (pj2[col] > upper)]
    print(f"{col}: Q1 = {q1}, Q3 = {q3}, IQR = {iqr}, fences = ({lower}, {upper}), flagged = {len(flagged)}")
```

IQR method was selected for outlier detection as it does not assume a bell-shaped distribution, and the columns are small whole numbers with a fixed menu. Fences are `Q1 - 1.5 * IQR` and `Q3 + 1.5 * IQR`.

And a flag means "unusual" not "wrong".

**Anomaly Report:** Total Spent differs from Price Per Unit in N rows with the largest gap being `16.0`. These rows were flagged not corrected

```python
nulls_before = raw.isna().sum().sum() + raw.isin(['UNKNOWN', 'ERROR']).sum().sum()

summary = pd.DataFrame({
    'Metric': ['Row count', 'Null count', 'Duplicate  rows'],
    'Before': [len(raw), nulls_before, raw.duplicated().sum()],
    'After': [len(pj2), pj2.isna().sum().sum(), pj2.duplicated().sum()]
})

summary
```

- **Row count:** 460 rows (4.6%) were removed, all because Transaction Date was missing and could not be reliably recovered
- **Null count:** this counts missing *cells* not rows
- **Duplicates:** none found before or after, so no rows were removed for this reason
